```python
# ============================================================================
# 11_quality_stability.ipynb
# RQ5: do existing quality metrics detect content variability?
# Scores every explanation with an LLM-as-a-judge protocol, then tests whether
# quality scores track the content distances computed in 08. The case of
# interest is the pair that scores equally well on quality while diverging in
# what it actually says.
# ============================================================================


# [Cell 1] Imports and global configuration
import os
import json
import time
import warnings
import threading
from pathlib import Path
from itertools import combinations
from collections import defaultdict
from concurrent.futures import ThreadPoolExecutor, as_completed

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm.auto import tqdm
from scipy.stats import spearmanr, pearsonr

from openai import OpenAI

warnings.filterwarnings("ignore")

SEED = 42
rng_global = np.random.default_rng(SEED)

ROOT = Path("..")
DATA_DIR = ROOT / "data"
FIG_DIR = ROOT / "results" / "figures"
TAB_DIR = ROOT / "results" / "tables"
LOG_DIR = ROOT / "logs"

for d in [DATA_DIR, FIG_DIR, TAB_DIR, LOG_DIR]:
    d.mkdir(parents=True, exist_ok=True)

JUDGE_DIR = DATA_DIR / "judge_cache"
JUDGE_DIR.mkdir(parents=True, exist_ok=True)

# API keys are read from a .env file at the project root rather than being
# written into the notebook, so that the code can be published unchanged.
from dotenv import load_dotenv, find_dotenv

_env = find_dotenv(usecwd=True)
if not _env:
    raise FileNotFoundError("No .env found at the project root")
load_dotenv(_env)
print(f"Loaded credentials from {_env}")


# [Cell 2] Plotting style: grayscale, publication settings
sns.set_theme(style="whitegrid", context="paper")
sns.set_palette(sns.color_palette("gray", n_colors=8))

plt.rcParams.update({
    "figure.dpi": 120,
    "savefig.dpi": 600,
    "font.family": "sans-serif",
    "axes.grid": True,
    "grid.color": "0.85",
    "grid.linewidth": 0.5,
    "axes.edgecolor": "0.3",
    "axes.labelcolor": "0.15",
    "text.color": "0.15",
    "xtick.color": "0.3",
    "ytick.color": "0.3",
    "image.cmap": "gray",
    "savefig.bbox": "tight",
    "savefig.transparent": False,
})


def save_fig(fig, name):
    """Save a figure as both PNG and PDF at 600 dpi. No caption is added."""
    for ext in ["png", "pdf"]:
        fig.savefig(FIG_DIR / f"{name}.{ext}", dpi=600, bbox_inches="tight")
    print(f"Saved: {name}.png / {name}.pdf")


def save_table(df, name, index=False):
    """Save a table as CSV under results/tables."""
    df.to_csv(TAB_DIR / f"{name}.csv", index=index, encoding="utf-8-sig")
    print(f"Saved: {name}.csv")


# [Cell 3] Judging configuration
# Two judges. Separating them by vendor is the standard mitigation for
# self-preference bias, but only one API is available here, so they are
# separated by generation instead. That is a weaker control: a judge and a
# generator from the same lineage may still share preferences. Reported as a
# limitation, and the reason judge agreement is checked before anything else.
JUDGES = {
    "gpt-5.6-terra": {"vendor": "openai", "api_name": "gpt-5.6-terra"},
    "gpt-5.4-mini":  {"vendor": "openai", "api_name": "gpt-5.4-mini"},
}

# GPT-5.6 rejects an explicit temperature, so determinism is requested
# through seed, the control the vendor actually exposes.
JUDGE_TEMPERATURE = None
JUDGE_SEED = 42
JUDGE_MAX_TOKENS = 1000

# Judging every generation would cost as much as the corpus itself. A stratified
# subsample is scored instead, sized so that every contrast retains enough pairs
# for the correlation analysis.
N_JUDGE_CASES = 120           # cases sampled per block
JUDGE_BLOCKS = ["A", "B", "D", "D0"]

MAX_WORKERS = 8
QUALITY_DIMENSIONS = ["accuracy", "clarity", "usefulness", "completeness"]

openai_client = OpenAI(api_key=os.environ["OPENAI_API_KEY"], timeout=90)


# [Cell 4] Load corpus and distances
corpus = pd.read_parquet(DATA_DIR / "generation_corpus.parquet")
corpus["evidence"] = corpus["evidence_json"].apply(json.loads)

pairs = pd.read_parquet(DATA_DIR / "content_distances.parquet")
case_level = pd.read_parquet(DATA_DIR / "case_level_distances.parquet")
fidelity = pd.read_parquet(DATA_DIR / "fidelity.parquet")

with open(DATA_DIR / "validation_summary.json", encoding="utf-8") as f:
    validation = json.load(f)

RANK_PRIMARY = validation["rank_primary"]
METRICS = ["feature_distance", "direction_distance"] + (
    ["rank_distance"] if RANK_PRIMARY else []
)

print(f"Corpus {len(corpus):,}   pairs {len(pairs):,}")


# [Cell 5] Sample the generations to be judged
# Sampling is at the case level so that both members of every pair are scored.
# Judging only one member would make the pairwise quality comparison impossible.
sampled = []
for block in JUDGE_BLOCKS:
    sub = corpus[corpus["block"] == block]
    if len(sub) == 0:
        continue
    cases = sub["case_id"].unique()
    take = min(N_JUDGE_CASES, len(cases))
    picked = rng_global.choice(cases, size=take, replace=False)
    sampled.append(sub[sub["case_id"].isin(picked)])

judge_set = pd.concat(sampled, ignore_index=True)
print(f"Generations to judge: {len(judge_set):,}")
print(judge_set.groupby("block").size().to_string())


# [Cell 6] Judging prompt
# The rubric deliberately mirrors what an evaluation in this literature would
# use. It scores a single explanation in isolation, which is precisely the
# design feature this research question interrogates: an isolated score cannot
# register that a different generation for the same case said something else.
JUDGE_SYSTEM = """You are evaluating an explanation written for a non-technical reader.

You will see the factual information the writer was given, followed by the explanation they produced.

Score the explanation on four dimensions, each from 1 to 5.

accuracy      Does the explanation represent the supplied information correctly?
              Check that every factor it names appears in the supplied information,
              and that the stated direction of each factor matches.
clarity       Can a reader with no statistical background follow it?
completeness  Does it cover the factors that were supplied?
usefulness    Does it help the reader understand and act on the assessment?

Scoring guidance:
5  excellent, no reservations
4  good, minor reservations
3  adequate
2  weak
1  unacceptable

Do not reward length. Do not penalise a short explanation that is accurate and clear.
Return JSON only, with no surrounding text or code fences."""


def build_judge_prompt(row):
    """Render the evidence and the explanation for the judge."""
    lines = ["SUPPLIED INFORMATION", ""]
    lines.append(f"Assessed value: {row['predicted_probability']:.3f}")
    lines.append(f"Typical value: {row['base_value']:.3f}")
    lines.append("")
    lines.append("Factors, most influential first:")
    for d in row["evidence"]:
        verb = "raised" if d["direction"] == "increases" else "lowered"
        lines.append(f"  {d['rank']}. {d['feature']} — {verb} the assessment")

    lines += ["", "EXPLANATION", "", row["explanation"], ""]
    lines.append(
        'Return: {"accuracy": n, "clarity": n, "completeness": n, '
        '"usefulness": n, "note": "one short sentence"}'
    )
    return "\n".join(lines)


# [Cell 7] Judge dispatch with caching
_lock = threading.Lock()
judge_stats = defaultdict(int)


def judge_key(row, judge):
    """Cache identifier for one generation scored by one judge."""
    return "_".join([
        judge, row["block"], row["domain"], row["model"], row["xai"],
        row["llm"], row["prompt"], f"{row['temperature']:.1f}",
        str(row["case_id"]), str(row["repeat"]),
    ])


def call_judge(judge, prompt):
    """Single judging call. All judges are OpenAI models."""
    spec = JUDGES[judge]
    kwargs = dict(
        model=spec["api_name"],
        max_completion_tokens=JUDGE_MAX_TOKENS,
        response_format={"type": "json_object"},
        seed=JUDGE_SEED,
        messages=[
            {"role": "system", "content": JUDGE_SYSTEM},
            {"role": "user", "content": prompt},
        ],
    )
    if JUDGE_TEMPERATURE is not None:
        kwargs["temperature"] = JUDGE_TEMPERATURE

    last_error = None
    for attempt in range(3):
        try:
            resp = openai_client.chat.completions.create(**kwargs)
            return json.loads(resp.choices[0].message.content)
        except Exception as exc:
            last_error = f"{type(exc).__name__}: {exc}"
            if attempt < 2:
                time.sleep(2 ** attempt)

    print(f"  judge call failed: {last_error}")
    return None


def judge_one(row, judge):
    """Score one generation with one judge, reusing the cache when present."""
    key = judge_key(row, judge)
    path = JUDGE_DIR / f"{key}.json"
    if path.exists():
        with _lock:
            judge_stats["cached"] += 1
        return json.loads(path.read_text(encoding="utf-8"))

    scores = call_judge(judge, build_judge_prompt(row))
    if scores is None:
        with _lock:
            judge_stats["failed"] += 1
        return None

    record = {
        "judge": judge,
        "block": row["block"], "domain": row["domain"], "model": row["model"],
        "xai": row["xai"], "llm": row["llm"], "prompt": row["prompt"],
        "temperature": float(row["temperature"]),
        "case_id": int(row["case_id"]), "repeat": int(row["repeat"]),
    }
    for dim in QUALITY_DIMENSIONS:
        val = scores.get(dim)
        record[dim] = float(val) if isinstance(val, (int, float)) else np.nan
    record["note"] = str(scores.get("note", ""))[:200]

    path.write_text(json.dumps(record, ensure_ascii=False, indent=1), encoding="utf-8")
    with _lock:
        judge_stats["scored"] += 1
    return record


tasks = [
    (row, judge)
    for row in judge_set.to_dict(orient="records")
    for judge in JUDGES
]
pending = [t for t in tasks if not (JUDGE_DIR / f"{judge_key(t[0], t[1])}.json").exists()]
print(f"Judging tasks pending: {len(pending):,} of {len(tasks):,}")

if pending:
    with ThreadPoolExecutor(max_workers=MAX_WORKERS) as pool:
        futures = [pool.submit(judge_one, row, judge) for row, judge in pending]
        for fut in tqdm(as_completed(futures), total=len(futures), desc="judging"):
            fut.result()

judge_records = []
for row, judge in tasks:
    path = JUDGE_DIR / f"{judge_key(row, judge)}.json"
    if path.exists():
        judge_records.append(json.loads(path.read_text(encoding="utf-8")))

judgements = pd.DataFrame(judge_records)
print(f"Judgements: {len(judgements):,}  {dict(judge_stats)}")


# [Cell 8] Judge agreement
# Establishes whether the quality construct is stable before it is used to test
# anything else. Poor agreement would mean the null result in RQ5 could not be
# distinguished from noise in the measurement.
key_cols = ["block", "domain", "model", "xai", "llm", "prompt",
            "temperature", "case_id", "repeat"]

wide = judgements.pivot_table(
    index=key_cols, columns="judge", values=QUALITY_DIMENSIONS
).reset_index()

judge_names = list(JUDGES)
agree_rows = []
for dim in QUALITY_DIMENSIONS:
    a = wide[(dim, judge_names[0])]
    b = wide[(dim, judge_names[1])]
    mask = a.notna() & b.notna()
    if mask.sum() < 30:
        continue
    agree_rows.append({
        "Dimension": dim,
        "N": int(mask.sum()),
        "Pearson r": round(float(pearsonr(a[mask], b[mask])[0]), 4),
        "Spearman rho": round(float(spearmanr(a[mask], b[mask])[0]), 4),
        "Mean abs diff": round(float(np.mean(np.abs(a[mask] - b[mask]))), 4),
        f"Mean {judge_names[0]}": round(float(a[mask].mean()), 3),
        f"Mean {judge_names[1]}": round(float(b[mask].mean()), 3),
    })

judge_agreement = pd.DataFrame(agree_rows)
save_table(judge_agreement, "tab54_judge_agreement")
display(judge_agreement)


# [Cell 9] Consolidated quality scores
# Scores are averaged across judges, as is standard practice for mitigating
# self-preference bias.
quality = (
    judgements.groupby(key_cols)[QUALITY_DIMENSIONS]
    .mean()
    .reset_index()
)
quality["quality_overall"] = quality[QUALITY_DIMENSIONS].mean(axis=1)
quality.to_parquet(DATA_DIR / "quality_scores.parquet", index=False)

quality_summary = (
    quality.groupby(["block", "llm", "prompt"])[
        QUALITY_DIMENSIONS + ["quality_overall"]
    ]
    .mean()
    .round(3)
    .reset_index()
)
save_table(quality_summary, "tab55_quality_by_condition")
display(quality_summary.head(12))


# [Cell 10] Pairwise quality differences
# For every pair already characterised by content distance, the corresponding
# difference in quality score is computed. The two quantities are then compared
# directly, which is the operational form of the research question.
def build_quality_pairs(quality_df, pairs_df):
    """Attach quality differences to the existing pair table."""
    q = quality_df.set_index([
        "block", "domain", "model", "xai", "llm", "prompt",
        "temperature", "case_id", "repeat",
    ])

    rows = []
    for _, p in tqdm(pairs_df.iterrows(), total=len(pairs_df), desc="matching"):
        key_a = (p["block"], p["domain"], p["a_model"], p["a_xai"], p["a_llm"],
                 p["a_prompt"], p["a_temp"], p["case_id"], p["a_repeat"])
        key_b = (p["block"], p["domain"], p["b_model"], p["b_xai"], p["b_llm"],
                 p["b_prompt"], p["b_temp"], p["case_id"], p["b_repeat"])

        if key_a not in q.index or key_b not in q.index:
            continue

        qa, qb = q.loc[key_a], q.loc[key_b]
        row = {
            "block": p["block"], "domain": p["domain"],
            "case_id": p["case_id"], "contrast": p["contrast"],
            "quality_a": float(qa["quality_overall"]),
            "quality_b": float(qb["quality_overall"]),
            "quality_diff": abs(float(qa["quality_overall"] - qb["quality_overall"])),
            "quality_mean": float((qa["quality_overall"] + qb["quality_overall"]) / 2),
            "accuracy_diff": abs(float(qa["accuracy"] - qb["accuracy"])),
            "accuracy_mean": float((qa["accuracy"] + qb["accuracy"]) / 2),
        }
        for m in METRICS:
            row[m] = p[m]
        rows.append(row)

    return pd.DataFrame(rows)


quality_pairs = build_quality_pairs(quality, pairs)
quality_pairs.to_parquet(DATA_DIR / "quality_distance_pairs.parquet", index=False)
print(f"Pairs with quality on both sides: {len(quality_pairs):,}")


# [Cell 11] Correlation between quality difference and content distance
# If quality metrics detected content variability, pairs that diverge in content
# would also diverge in score. A weak correlation is the evidence that they do
# not. Inference is by case-level cluster bootstrap, matching the primary
# analysis.
def bootstrap_correlation(df, x, y, B=1000, level=0.95, rng=None):
    """Spearman correlation with a case-level cluster bootstrap interval."""
    rng = rng or np.random.default_rng(SEED)
    valid = df[[x, y, "case_id"]].dropna()
    if valid["case_id"].nunique() < 20:
        return None

    point = float(spearmanr(valid[x], valid[y])[0])
    cases = valid["case_id"].unique()
    grouped = {c: g for c, g in valid.groupby("case_id")}

    stats = []
    for _ in range(B):
        picked = rng.choice(cases, size=len(cases), replace=True)
        resampled = pd.concat([grouped[c] for c in picked], ignore_index=True)
        if resampled[x].std() == 0 or resampled[y].std() == 0:
            continue
        stats.append(spearmanr(resampled[x], resampled[y])[0])

    if len(stats) < B * 0.5:
        return None

    alpha = (1 - level) / 2
    return {
        "rho": round(point, 4),
        "ci_low": round(float(np.quantile(stats, alpha)), 4),
        "ci_high": round(float(np.quantile(stats, 1 - alpha)), 4),
        "n_pairs": int(len(valid)),
        "n_cases": int(valid["case_id"].nunique()),
    }


corr_rows = []
rng = np.random.default_rng(SEED)

for block in quality_pairs["block"].unique():
    sub_block = quality_pairs[quality_pairs["block"] == block]
    for contrast in sub_block["contrast"].unique():
        sub = sub_block[sub_block["contrast"] == contrast]
        if len(sub) < 50:
            continue
        for metric in METRICS:
            for qcol, qlabel in [("quality_diff", "Overall quality"),
                                 ("accuracy_diff", "Accuracy")]:
                res = bootstrap_correlation(sub, metric, qcol, rng=rng)
                if res is None:
                    continue
                corr_rows.append({
                    "Block": block,
                    "Contrast": contrast,
                    "Distance metric": metric.replace("_", " "),
                    "Quality measure": qlabel,
                    "Spearman rho": res["rho"],
                    "CI low": res["ci_low"],
                    "CI high": res["ci_high"],
                    "N pairs": res["n_pairs"],
                    "Excludes zero": bool(res["ci_low"] > 0 or res["ci_high"] < 0),
                })

correlations = pd.DataFrame(corr_rows)
save_table(correlations, "tab56_quality_distance_correlation")

print("\nOverall association between content distance and quality difference:")
print(correlations.groupby("Quality measure")["Spearman rho"].describe().round(4).to_string())
display(correlations.head(15))


# [Cell 12] The blind spot: high quality, divergent content
# The central quantity for this research question. Pairs are classified by
# whether both explanations scored well and whether their content nonetheless
# differs substantially.
QUALITY_HIGH = 4.0            # both explanations judged good or better
DISTANCE_HIGH = 0.30          # substantial divergence in content

quality_pairs["both_high_quality"] = (
    (quality_pairs["quality_a"] >= QUALITY_HIGH)
    & (quality_pairs["quality_b"] >= QUALITY_HIGH)
)
quality_pairs["content_divergent"] = quality_pairs["feature_distance"] >= DISTANCE_HIGH
quality_pairs["direction_divergent"] = quality_pairs["direction_distance"] > 0

blind_rows = []
for (block, contrast), g in quality_pairs.groupby(["block", "contrast"]):
    if len(g) < 50:
        continue
    high_q = g[g["both_high_quality"]]
    blind_rows.append({
        "Block": block,
        "Contrast": contrast,
        "N pairs": len(g),
        "Both high quality": len(high_q),
        "Share high quality": round(len(high_q) / len(g), 4),
        "High quality and divergent": int(high_q["content_divergent"].sum()),
        "Blind spot share": round(
            float(high_q["content_divergent"].mean()) if len(high_q) else np.nan, 4
        ),
        "High quality with direction conflict": int(
            high_q["direction_divergent"].sum()
        ),
        "Direction conflict share": round(
            float(high_q["direction_divergent"].mean()) if len(high_q) else np.nan, 4
        ),
    })

blind_spot = pd.DataFrame(blind_rows)
save_table(blind_spot, "tab57_quality_blind_spot")

overall_blind = float(
    quality_pairs[quality_pairs["both_high_quality"]]["content_divergent"].mean()
)
overall_dir = float(
    quality_pairs[quality_pairs["both_high_quality"]]["direction_divergent"].mean()
)
print(f"\nAmong pairs where both explanations scored {QUALITY_HIGH} or above:")
print(f"  content divergent (feature distance >= {DISTANCE_HIGH}): {overall_blind:.4f}")
print(f"  at least one direction conflict:                        {overall_dir:.4f}")

display(blind_spot)


# [Cell 13] Quality is not lower where content diverges
# A stronger form of the same test. If quality metrics captured content
# variability, mean quality would be lower among divergent pairs.
strat_rows = []
for metric in METRICS:
    sub = quality_pairs[[metric, "quality_mean", "accuracy_mean", "case_id"]].dropna()
    if len(sub) < 100:
        continue

    # Quartiles of content distance
    cats = pd.qcut(sub[metric], 4, duplicates="drop")
    n_bins = len(cats.cat.categories)
    if n_bins < 2:
        print(f"  {metric}: too many tied values to stratify, skipped")
        continue
    sub["quartile"] = cats.cat.rename_categories(
        [f"Q{i + 1}" for i in range(n_bins)]
    )
    if n_bins < 4:
        print(f"  {metric}: only {n_bins} distinct bins after tie removal")
    for q, g in sub.groupby("quartile"):
        strat_rows.append({
            "Distance metric": metric.replace("_", " "),
            "Distance quartile": str(q),
            "Mean distance": round(float(g[metric].mean()), 4),
            "Mean quality": round(float(g["quality_mean"].mean()), 4),
            "Mean accuracy score": round(float(g["accuracy_mean"].mean()), 4),
            "N pairs": len(g),
        })

stratified = pd.DataFrame(strat_rows)
save_table(stratified, "tab58_quality_by_distance_quartile")

for metric in stratified["Distance metric"].unique():
    s = stratified[stratified["Distance metric"] == metric]
    drop = float(s["Mean quality"].iloc[0] - s["Mean quality"].iloc[-1])
    print(f"{metric}: quality change from lowest to highest distance quartile "
          f"= {drop:+.4f} points on a 5-point scale")

display(stratified)


# [Cell 14] Quality against measured fidelity
# A separate check on the judges themselves. The accuracy dimension asks the
# judge to verify that stated directions match the supplied evidence, which is
# what the fidelity metric measures mechanically. Weak agreement indicates that
# the judge is not performing the verification the rubric requests.
fid_keys = ["block", "domain", "model", "xai", "llm", "prompt",
            "temperature", "case_id", "repeat"]

merged_fid = quality.merge(fidelity, on=fid_keys, how="inner")
print(f"Generations with both quality and fidelity: {len(merged_fid):,}")

fid_rows = []
for qcol in ["accuracy", "quality_overall"]:
    for fcol in ["direction_agreement", "hallucination_rate", "omission_rate"]:
        sub = merged_fid[[qcol, fcol, "case_id"]].dropna()
        if len(sub) < 100 or sub[fcol].std() == 0:
            continue
        res = bootstrap_correlation(
            sub.rename(columns={qcol: "x", fcol: "y"}), "x", "y",
            rng=np.random.default_rng(SEED + 1),
        )
        if res is None:
            continue
        fid_rows.append({
            "Judge dimension": qcol,
            "Fidelity metric": fcol.replace("_", " "),
            "Spearman rho": res["rho"],
            "CI low": res["ci_low"],
            "CI high": res["ci_high"],
            "N": res["n_pairs"],
            "Excludes zero": bool(res["ci_low"] > 0 or res["ci_high"] < 0),
        })

judge_fidelity = pd.DataFrame(fid_rows)
save_table(judge_fidelity, "tab59_judge_versus_measured_fidelity")

acc_dir = judge_fidelity[
    (judge_fidelity["Judge dimension"] == "accuracy")
    & (judge_fidelity["Fidelity metric"] == "direction agreement")
]
if len(acc_dir):
    r = float(acc_dir["Spearman rho"].iloc[0])
    print(f"\nJudge accuracy score against measured direction agreement: rho = {r:.4f}")
    if abs(r) < 0.3:
        print("The judges' accuracy scores track measured direction fidelity only "
              "weakly, despite the rubric instructing them to verify directions. "
              "This is a substantive finding about LLM-as-a-judge evaluation.")

display(judge_fidelity)


# [Cell 15] Figure: quality difference against content distance
fig, axes = plt.subplots(1, len(METRICS), figsize=(3.2 * len(METRICS), 3.2))
if len(METRICS) == 1:
    axes = [axes]

for ax, metric in zip(axes, METRICS):
    sub = quality_pairs[[metric, "quality_diff"]].dropna()
    if len(sub) > 4000:
        sub = sub.sample(4000, random_state=SEED)

    ax.scatter(
        sub[metric], sub["quality_diff"],
        s=5, color="0.55", alpha=0.35, edgecolor="none", zorder=2,
    )

    # Binned means make the absence of a trend legible against the scatter
    bins = np.linspace(0, 1, 11)
    centres, means = [], []
    for lo, hi in zip(bins[:-1], bins[1:]):
        m = sub[(sub[metric] >= lo) & (sub[metric] < hi)]
        if len(m) >= 20:
            centres.append((lo + hi) / 2)
            means.append(m["quality_diff"].mean())
    if centres:
        ax.plot(centres, means, color="0.1", linewidth=1.4,
                marker="o", markersize=4, zorder=3)

    ax.set_xlabel(metric.replace("_", " "))
    ax.set_ylabel("Quality difference" if metric == METRICS[0] else "")
    ax.set_xlim(-0.02, 1.02)
    ax.set_ylim(-0.05, max(2.0, sub["quality_diff"].quantile(0.99)))

plt.tight_layout()
save_fig(fig, "fig36_quality_versus_distance")
plt.show()


# [Cell 16] Figure: the blind spot region
fig, ax = plt.subplots(figsize=(5.0, 4.0))

sub = quality_pairs[["feature_distance", "quality_mean"]].dropna()
if len(sub) > 5000:
    sub = sub.sample(5000, random_state=SEED)

ax.scatter(
    sub["feature_distance"], sub["quality_mean"],
    s=6, color="0.6", alpha=0.3, edgecolor="none", zorder=2,
)

# Shade the region where explanations score well yet say different things
ax.axhspan(QUALITY_HIGH, 5.05, xmin=DISTANCE_HIGH, xmax=1.0,
           color="0.25", alpha=0.16, zorder=1)
ax.axhline(QUALITY_HIGH, color="0.1", linestyle="--", linewidth=0.9, zorder=3)
ax.axvline(DISTANCE_HIGH, color="0.1", linestyle="--", linewidth=0.9, zorder=3)

in_region = (
    (quality_pairs["feature_distance"] >= DISTANCE_HIGH)
    & (quality_pairs["quality_mean"] >= QUALITY_HIGH)
).sum()
share = in_region / quality_pairs["feature_distance"].notna().sum()

ax.text(
    0.97, 4.92, f"{share:.1%} of pairs", ha="right", va="top",
    fontsize=8, color="0.15",
    bbox=dict(boxstyle="round,pad=0.3", facecolor="white",
              edgecolor="0.5", linewidth=0.6),
)

ax.set_xlabel("Feature distance")
ax.set_ylabel("Mean quality score")
ax.set_xlim(-0.02, 1.02)
ax.set_ylim(1.0, 5.1)

plt.tight_layout()
save_fig(fig, "fig37_quality_stability_blind_spot")
plt.show()


# [Cell 17] Figure: quality by content distance quartile
if len(stratified):
    fig, ax = plt.subplots(figsize=(5.6, 3.2))

    metrics_present = stratified["Distance metric"].unique()
    quartiles = ["Q1", "Q2", "Q3", "Q4"]
    x = np.arange(len(quartiles))
    width = 0.8 / max(len(metrics_present), 1)
    shades = ["0.25", "0.50", "0.72"]

    for i, metric in enumerate(metrics_present):
        s = stratified[stratified["Distance metric"] == metric].set_index(
            "Distance quartile"
        ).reindex(quartiles)
        ax.bar(
            x + (i - (len(metrics_present) - 1) / 2) * width,
            s["Mean quality"], width,
            color=shades[i % 3], edgecolor="0.2", linewidth=0.7, label=metric,
        )

    ax.set_xticks(x)
    ax.set_xticklabels(["Q1\nlowest", "Q2", "Q3", "Q4\nhighest"], fontsize=8)
    ax.set_xlabel("Content distance quartile")
    ax.set_ylabel("Mean quality score")
    ax.set_ylim(1, 5.2)
    ax.legend(fontsize=7, frameon=True, loc="lower left")

    plt.tight_layout()
    save_fig(fig, "fig38_quality_by_distance_quartile")
    plt.show()


# [Cell 18] Figure: judge accuracy against measured fidelity
merged_plot = merged_fid[["accuracy", "direction_agreement"]].dropna()
if len(merged_plot) >= 100:
    fig, ax = plt.subplots(figsize=(4.6, 3.4))

    bins = np.linspace(0, 1, 11)
    centres, means, counts = [], [], []
    for lo, hi in zip(bins[:-1], bins[1:]):
        m = merged_plot[
            (merged_plot["direction_agreement"] >= lo)
            & (merged_plot["direction_agreement"] < hi)
        ]
        if len(m) >= 10:
            centres.append((lo + hi) / 2)
            means.append(m["accuracy"].mean())
            counts.append(len(m))

    if centres:
        ax.plot(centres, means, color="0.2", linewidth=1.4,
                marker="o", markersize=5, zorder=3)
        for c, m, n in zip(centres, means, counts):
            ax.annotate(f"n={n}", (c, m), textcoords="offset points",
                        xytext=(0, 7), ha="center", fontsize=6, color="0.45")

    ax.set_xlabel("Measured direction agreement")
    ax.set_ylabel("Judge accuracy score")
    ax.set_xlim(-0.05, 1.05)
    ax.set_ylim(1, 5.2)

    plt.tight_layout()
    save_fig(fig, "fig39_judge_accuracy_versus_fidelity")
    plt.show()


# [Cell 19] Answer to RQ5
answer = {
    "question": "Do existing quality metrics detect content variability?",
    "judge_agreement_overall": (
        round(float(judge_agreement["Spearman rho"].mean()), 4)
        if len(judge_agreement) else None
    ),
    "median_correlation_distance_quality": (
        round(float(correlations[
            correlations["Quality measure"] == "Overall quality"
        ]["Spearman rho"].median()), 4)
        if len(correlations) else None
    ),
    "median_correlation_distance_accuracy": (
        round(float(correlations[
            correlations["Quality measure"] == "Accuracy"
        ]["Spearman rho"].median()), 4)
        if len(correlations) else None
    ),
    "blind_spot_share": round(overall_blind, 4),
    "direction_conflict_among_high_quality": round(overall_dir, 4),
    "quality_drop_across_distance_quartiles": (
        round(float(
            stratified.groupby("Distance metric")["Mean quality"]
            .agg(lambda s: s.iloc[0] - s.iloc[-1]).mean()
        ), 4) if len(stratified) else None
    ),
    "judge_accuracy_vs_measured_fidelity": (
        round(float(acc_dir["Spearman rho"].iloc[0]), 4) if len(acc_dir) else None
    ),
}

rq5 = pd.DataFrame([answer]).T.reset_index()
rq5.columns = ["Quantity", "Value"]
save_table(rq5, "tab60_rq5_answer")

print(json.dumps(answer, indent=2))

med_corr = answer["median_correlation_distance_quality"]
if med_corr is not None and abs(med_corr) < 0.30:
    print("""
Quality scores and content distance are only weakly associated. Explanations
that name different drivers, or state opposite directions for the same driver,
receive comparable quality scores. Quality-only evaluation therefore does not
register the variability this study measures.""")


# [Cell 20] Persist the manifest
manifest = {
    "notebook": "11_quality_stability",
    "timestamp": pd.Timestamp.now().isoformat(),
    "seed": SEED,
    "judges": JUDGES,
    "judge_temperature": JUDGE_TEMPERATURE,
    "judge_seed": JUDGE_SEED,
    "quality_dimensions": QUALITY_DIMENSIONS,
    "n_generations_judged": int(len(quality)),
    "n_judgements": int(len(judgements)),
    "n_quality_pairs": int(len(quality_pairs)),
    "thresholds": {"quality_high": QUALITY_HIGH, "distance_high": DISTANCE_HIGH},
    "judge_agreement": judge_agreement.to_dict(orient="records"),
    "correlations": correlations.to_dict(orient="records"),
    "blind_spot": blind_spot.to_dict(orient="records"),
    "stratified": stratified.to_dict(orient="records"),
    "judge_vs_fidelity": judge_fidelity.to_dict(orient="records"),
    "rq5_answer": answer,
}

with open(DATA_DIR / "quality_manifest.json", "w", encoding="utf-8") as f:
    json.dump(manifest, f, ensure_ascii=False, indent=2, default=str)
print("Saved: quality_manifest.json")

stamp = pd.Timestamp.now().strftime("%Y%m%d")
with open(LOG_DIR / f"11_run_{stamp}.json", "w", encoding="utf-8") as f:
    json.dump(manifest, f, ensure_ascii=False, indent=2, default=str)
print(f"Saved: logs/11_run_{stamp}.json")


# [Cell 21] Study-wide summary
# Consolidates the answers to every research question into one table, so that
# the results section can be written directly from a single artefact.
manifest_paths = {
    "corpus": "corpus_manifest.json",
    "distance": "distance_manifest.json",
    "delta": "delta_manifest.json",
    "robustness": "robustness_manifest.json",
    "quality": "quality_manifest.json",
}
manifests = {}
for name, fn in manifest_paths.items():
    p = DATA_DIR / fn
    if p.exists():
        manifests[name] = json.loads(p.read_text(encoding="utf-8"))

summary_rows = []

if "delta" in manifests:
    for a in manifests["delta"].get("rq_answers", []):
        summary_rows.append({
            "RQ": a["RQ"],
            "Finding": a["Question"],
            "Value": a["Delta"],
            "Interval": a["CI"],
            "Source": "tab46",
        })

summary_rows.append({
    "RQ": "RQ5",
    "Finding": "Association between content distance and quality difference",
    "Value": answer["median_correlation_distance_quality"],
    "Interval": "see tab56",
    "Source": "tab56",
})
summary_rows.append({
    "RQ": "RQ5",
    "Finding": "Divergent content among pairs both judged 4.0 or above",
    "Value": answer["blind_spot_share"],
    "Interval": "see tab57",
    "Source": "tab57",
})

study_summary = pd.DataFrame(summary_rows)
save_table(study_summary, "tab61_study_summary")
display(study_summary)


# [Cell 22] Verification and handover notes
print("Outputs:")
for p in sorted(TAB_DIR.glob("tab5[4-9]*")) + sorted(TAB_DIR.glob("tab6[0-1]*")) \
        + sorted(FIG_DIR.glob("fig3[6-9]*")):
    print(f"  {p.relative_to(ROOT)}")

print(f"""
RQ5 summary
-----------
Generations judged            {len(quality):,}
Judge agreement (mean rho)    {answer['judge_agreement_overall']}
Distance-quality correlation  {answer['median_correlation_distance_quality']}
Blind spot share              {answer['blind_spot_share']}

Reporting notes
  - The judging protocol scores each explanation in isolation, which is the
    prevailing practice in this literature. That isolation is the mechanism
    behind the finding, not a flaw in the implement
    2ation: a score assigned to
    one explanation cannot register that another generation for the same case
    said something different.
  - Both judges are themselves LLMs. The result therefore concerns
    LLM-as-a-judge evaluation specifically, and does not establish what human
    raters would detect. State this limitation explicitly.
  - The thresholds for high quality and substantial divergence are reporting
    conventions, not inferential cut-offs. The continuous relationship in
    tab58 and fig36 carries the argument; the thresholded share in tab57 is an
    illustration of it.

All analysis is complete. Study artefacts:
  data/*.parquet          analysis-ready datasets
  data/*_manifest.json    per-stage provenance
  results/figures/        39 figures, PNG and PDF at 600 dpi
  results/tables/         61 tables
  logs/                   run records for every notebook
""")
```
