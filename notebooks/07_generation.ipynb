```python
# ============================================================================
# 07_generation.ipynb
# Main experiment: generate the full explanation corpus
# Enumerates every condition in Blocks A, A', B, C, D and D-0, then generates
# the narratives with per-condition caching so that an interrupted run resumes
# without re-issuing completed calls. This notebook produces the raw material
# for all subsequent analysis and is the most expensive step in the study.
# ============================================================================


# [Cell 1] Imports and global configuration
import os
import json
import time
import random
import hashlib
import warnings
import threading
from pathlib import Path
from collections import Counter, defaultdict
from concurrent.futures import ThreadPoolExecutor, as_completed

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm.auto import tqdm

from openai import OpenAI

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
random.seed(SEED)

ROOT = Path("..")
DATA_DIR = ROOT / "data"
PROMPT_DIR = ROOT / "prompts"
GEN_DIR = DATA_DIR / "generations"
FIG_DIR = ROOT / "results" / "figures"
TAB_DIR = ROOT / "results" / "tables"
LOG_DIR = ROOT / "logs"

for d in [DATA_DIR, PROMPT_DIR, GEN_DIR, FIG_DIR, TAB_DIR, LOG_DIR]:
    d.mkdir(parents=True, exist_ok=True)

# API keys are read from a .env file at the project root rather than being
# written into the notebook, so that the code can be published unchanged.
from dotenv import load_dotenv, find_dotenv

_env = find_dotenv(usecwd=True)
if not _env:
    raise FileNotFoundError("No .env found at the project root")
load_dotenv(_env)
print(f"Loaded credentials from {_env}")


# [Cell 2] Plotting style: grayscale, publication settings
sns.set_theme(style="whitegrid", context="paper")
sns.set_palette(sns.color_palette("gray", n_colors=8))

plt.rcParams.update({
    "figure.dpi": 120,
    "savefig.dpi": 600,
    "font.family": "sans-serif",
    "axes.grid": True,
    "grid.color": "0.85",
    "grid.linewidth": 0.5,
    "axes.edgecolor": "0.3",
    "axes.labelcolor": "0.15",
    "text.color": "0.15",
    "xtick.color": "0.3",
    "ytick.color": "0.3",
    "image.cmap": "gray",
    "savefig.bbox": "tight",
    "savefig.transparent": False,
})


def save_fig(fig, name):
    """Save a figure as both PNG and PDF at 600 dpi. No caption is added."""
    for ext in ["png", "pdf"]:
        fig.savefig(FIG_DIR / f"{name}.{ext}", dpi=600, bbox_inches="tight")
    print(f"Saved: {name}.png / {name}.pdf")


def save_table(df, name, index=False):
    """Save a table as CSV under results/tables."""
    df.to_csv(TAB_DIR / f"{name}.csv", index=index, encoding="utf-8-sig")
    print(f"Saved: {name}.csv")


# [Cell 3] Load the validated design
with open(DATA_DIR / "design_recommendation.json", encoding="utf-8") as f:
    design = json.load(f)

with open(DATA_DIR / "validation_summary.json", encoding="utf-8") as f:
    validation = json.load(f)

if not validation["proceed"]:
    raise RuntimeError(
        "Claim extraction did not meet the pre-registered gates in 05. "
        "Revise the extractors before generating the full corpus."
    )

FINAL = design["final"]
print("Design in force:")
for k, v in FINAL.items():
    print(f"  {k}: {v}")


# [Cell 4] Generator configuration
# Three generators spanning two generations and two capability tiers.
#
#                 mid            budget
#   GPT-5.6       terra          luna
#   GPT-5.4        -             5.4-mini
#
# terra vs luna isolates the tier difference within a generation; luna vs
# 5.4-mini isolates the generation difference at a comparable tier.
#
# All three are OpenAI models. Vendor diversity would be preferable, since a
# single-vendor panel confounds the LLM factor with a shared training lineage,
# but only one API is available for this study. This is stated as a limitation
# rather than worked around.
LLMS = {
    "gpt-5.6-terra": {"vendor": "openai", "api_name": "gpt-5.6-terra",
                      "generation": "5.6", "tier": "mid"},
    "gpt-5.6-luna":  {"vendor": "openai", "api_name": "gpt-5.6-luna",
                      "generation": "5.6", "tier": "budget"},
    "gpt-5.4-mini":  {"vendor": "openai", "api_name": "gpt-5.4-mini",
                      "generation": "5.4", "tier": "budget"},
}

REFERENCE_LLM = "gpt-5.4-mini"    # held fixed in Blocks A, A', D, D-0
REFERENCE_PROMPT = "basic"        # held fixed in Blocks A, A', C, D, D-0
REFERENCE_MODEL = "XGBoost"       # held fixed in Block B
REFERENCE_XAI = "SHAP"            # held fixed in Block B

PROMPTS = ["basic", "constrained", "verification"]
MODELS = ["XGBoost", "LightGBM", "RandomForest"]
XAI_METHODS = ["SHAP", "LIME"]

# GPT-5.6 accepts only its default temperature, so 1.0 is used throughout and
# the deterministic condition is created with a fixed seed instead. Seed is
# the reproducibility control the vendor actually exposes; temperature merely
# suppresses sampling. RQ6 therefore asks whether the vendor's own determinism
# mechanism delivers content reproducibility.
TEMPERATURE_MAIN = 1.0
SEED_FIXED = 42               # Block D-0
SEED_FREE = None              # every other block
MAX_TOKENS = 400

# Execution controls
MAX_WORKERS = 8                   # concurrent requests
MAX_RETRIES = 5
BASE_BACKOFF = 2.0                # seconds, doubled on each retry
REQUEST_TIMEOUT = 90

openai_client = OpenAI(api_key=os.environ["OPENAI_API_KEY"], timeout=REQUEST_TIMEOUT)


# [Cell 5] Load evidence blocks and case registry
with open(DATA_DIR / "evidence_blocks.json", encoding="utf-8") as f:
    evidence_blocks = json.load(f)

case_registry = pd.read_parquet(DATA_DIR / "case_registry.parquet")

# Index evidence blocks for constant-time lookup during enumeration
block_index = {
    (b["domain"], b["model"], b["xai"], b["case_id"]): b for b in evidence_blocks
}

case_sets = {
    (row["domain"], row["case_set"]): sorted(
        case_registry[
            (case_registry["domain"] == row["domain"])
            & (case_registry["case_set"] == row["case_set"])
        ]["case_id"].unique().tolist()
    )
    for _, row in case_registry[["domain", "case_set"]].drop_duplicates().iterrows()
}

for k, v in case_sets.items():
    print(f"  {k}: {len(v)} cases")


# [Cell 6] Prompt loading
PROMPT_TEXT = {
    name: (PROMPT_DIR / f"{name}_v1.txt").read_text(encoding="utf-8")
    for name in PROMPTS
}
PROMPT_HASH = {
    name: hashlib.md5(text.encode()).hexdigest()[:12]
    for name, text in PROMPT_TEXT.items()
}
print("Prompt versions in use:")
for name, h in PROMPT_HASH.items():
    print(f"  {name}: {h}")


# [Cell 7] User message construction
# Identical to the pilot in 04, so that parser development transfers directly.
DOMAIN_FRAMING = {
    "credit": {
        "subject": "a loan applicant",
        "outcome": "the likelihood that the loan will not be repaid",
    },
    "churn": {
        "subject": "a subscriber",
        "outcome": "the likelihood that the subscriber will leave the service",
    },
}


def build_user_message(block):
    """Render the structured evidence block as the user-facing prompt text."""
    framing = DOMAIN_FRAMING[block["domain"]]
    lines = [
        f"Subject: {framing['subject']}",
        f"Assessment: {framing['outcome']}",
        f"Assessed value: {block['predicted_probability']:.3f} "
        f"(typical value for this population: {block['base_value']:.3f})",
        "",
        "Factors that drove this assessment, most influential first:",
    ]
    for d in block["drivers"]:
        verb = "raised" if d["direction"] == "increases" else "lowered"
        lines.append(
            f"  {d['rank']}. {d['feature']} — {verb} the assessment "
            f"(relative weight {abs(d['contribution_norm']):.2f})"
        )
    return "\n".join(lines)


# [Cell 8] Condition enumeration
# Every generation in the study is described by a single condition record.
# Enumerating them up front makes the corpus auditable: the plan can be checked
# against the design before a single call is made.
def make_condition(block_name, domain, model, xai, llm, prompt,
                   temperature, case_id, repeat, seed=SEED_FREE):
    """Assemble one condition record."""
    return {
        "block": block_name,
        "domain": domain,
        "model": model,
        "xai": xai,
        "llm": llm,
        "prompt": prompt,
        "temperature": temperature,
        "seed": seed,
        "case_id": int(case_id),
        "repeat": int(repeat),
    }


def enumerate_conditions():
    """Build the full list of conditions across all six blocks."""
    conds = []

    # Block A: model and XAI vary, generator and prompt fixed, primary domain
    for case_id in case_sets[("credit", "main")]:
        for model in MODELS:
            for xai in XAI_METHODS:
                for rep in range(1, FINAL["n_repeats_main"] + 1):
                    conds.append(make_condition(
                        "A", "credit", model, xai,
                        REFERENCE_LLM, REFERENCE_PROMPT,
                        TEMPERATURE_MAIN, case_id, rep,
                    ))

    # Block A': identical structure in the control domain
    for case_id in case_sets[("churn", "main")]:
        for model in MODELS:
            for xai in XAI_METHODS:
                for rep in range(1, FINAL["n_repeats_main"] + 1):
                    conds.append(make_condition(
                        "A_prime", "churn", model, xai,
                        REFERENCE_LLM, REFERENCE_PROMPT,
                        TEMPERATURE_MAIN, case_id, rep,
                    ))

    # Block B: generator and prompt vary, evidence held fixed
    for case_id in case_sets[("credit", "main")]:
        for llm in LLMS:
            for prompt in PROMPTS:
                for rep in range(1, FINAL["n_repeats_main"] + 1):
                    conds.append(make_condition(
                        "B", "credit", REFERENCE_MODEL, REFERENCE_XAI,
                        llm, prompt, TEMPERATURE_MAIN, case_id, rep,
                    ))

    # Block C: model, XAI and generator vary together on a reduced case set
    for case_id in case_sets[("credit", "bridge")]:
        for model in MODELS:
            for xai in XAI_METHODS:
                for llm in LLMS:
                    for rep in range(1, 3):
                        conds.append(make_condition(
                            "C", "credit", model, xai, llm, REFERENCE_PROMPT,
                            TEMPERATURE_MAIN, case_id, rep,
                        ))

    # Block D: nothing varies except the repetition index
    for case_id in case_sets[("credit", "deep")]:
        for llm in LLMS:
            for rep in range(1, FINAL["n_repeats_deep"] + 1):
                conds.append(make_condition(
                    "D", "credit", REFERENCE_MODEL, REFERENCE_XAI,
                    llm, REFERENCE_PROMPT, TEMPERATURE_MAIN, case_id, rep,
                ))

    # Block D-0: Block D repeated with a fixed seed. Everything else is
    # identical, so the contrast isolates the vendor determinism control.
    for case_id in case_sets[("credit", "deep")]:
        for llm in LLMS:
            for rep in range(1, FINAL["n_repeats_deep"] + 1):
                conds.append(make_condition(
                    "D0", "credit", REFERENCE_MODEL, REFERENCE_XAI,
                    llm, REFERENCE_PROMPT, TEMPERATURE_MAIN, case_id, rep,
                    seed=SEED_FIXED,
                ))

    return conds


conditions = enumerate_conditions()
plan = pd.DataFrame(conditions)

plan_summary = (
    plan.groupby("block")
    .agg(
        N=("case_id", "size"),
        cases=("case_id", "nunique"),
        models=("model", "nunique"),
        xai=("xai", "nunique"),
        llms=("llm", "nunique"),
        prompts=("prompt", "nunique"),
        repeats=("repeat", "nunique"),
    )
    .reset_index()
)
plan_summary.loc[len(plan_summary)] = ["Total", len(plan), *[None] * 6]

save_table(plan_summary, "tab27_generation_plan")
print(f"\nTotal generations planned: {len(plan):,}")
display(plan_summary)


# [Cell 9] Cache addressing
# The cache key covers every factor that could change the output. Two runs with
# identical keys are interchangeable; any change to a factor produces a new key,
# so stale results can never be silently reused.
def condition_key(cond):
    """Deterministic identifier for a condition, used as the cache filename."""
    raw = "|".join([
        cond["block"], cond["domain"], cond["model"], cond["xai"],
        cond["llm"], cond["prompt"], f"{cond['temperature']:.2f}",
        f"seed={cond['seed']}",
        str(cond["case_id"]), str(cond["repeat"]),
        PROMPT_HASH[cond["prompt"]],
    ])
    return hashlib.md5(raw.encode()).hexdigest()[:20]


def condition_path(cond):
    """Filesystem location for a condition's cached output."""
    block_dir = GEN_DIR / cond["block"].lower()
    block_dir.mkdir(parents=True, exist_ok=True)
    return block_dir / f"{condition_key(cond)}.json"


for cond in conditions:
    cond["_key"] = condition_key(cond)
    cond["_path"] = condition_path(cond)

# Guard against key collisions, which would silently overwrite results
keys = [c["_key"] for c in conditions]
if len(set(keys)) != len(keys):
    dup = [k for k, n in Counter(keys).items() if n > 1]
    raise RuntimeError(f"Cache key collision on {len(dup)} keys. Widen the key.")
print(f"Cache keys unique: {len(set(keys)):,}")


# [Cell 10] API call wrappers
_usage_lock = threading.Lock()
usage_tally = defaultdict(lambda: {"calls": 0, "input_tokens": 0, "output_tokens": 0})


def call_openai(api_name, system, user, temperature, seed=None):
    """Single OpenAI chat completion.

    system_fingerprint is returned alongside the text. The vendor documents
    seed as best-effort: identical seeds reproduce outputs only while the
    backend is unchanged, and a change in fingerprint is the signal that the
    guarantee lapsed. Recording it makes that visible in the analysis.
    """
    kwargs = dict(
        model=api_name,
        max_completion_tokens=MAX_TOKENS,
        messages=[
            {"role": "system", "content": system},
            {"role": "user", "content": user},
        ],
    )
    if temperature is not None:
        kwargs["temperature"] = temperature
    if seed is not None:
        kwargs["seed"] = seed

    resp = openai_client.chat.completions.create(**kwargs)
    return (
        resp.choices[0].message.content.strip(),
        resp.usage.prompt_tokens,
        resp.usage.completion_tokens,
        getattr(resp, "system_fingerprint", None),
    )


def call_with_retry(cond, system, user):
    """Call the generator with exponential backoff. One vendor, so no dispatch."""
    spec = LLMS[cond["llm"]]

    last_error = None
    for attempt in range(MAX_RETRIES):
        try:
            text, tin, tout, fingerprint = call_openai(
                spec["api_name"], system, user,
                cond["temperature"], cond.get("seed"),
            )
            with _usage_lock:
                u = usage_tally[cond["llm"]]
                u["calls"] += 1
                u["input_tokens"] += tin
                u["output_tokens"] += tout
            return text, tin, tout, fingerprint, None
        except Exception as exc:
            last_error = f"{type(exc).__name__}: {exc}"
            if attempt < MAX_RETRIES - 1:
                # Jitter prevents synchronised retries across worker threads
                delay = BASE_BACKOFF * (2 ** attempt) * (0.5 + random.random())
                time.sleep(delay)

    return None, 0, 0, None, last_error


# [Cell 11] Single-condition worker
def run_condition(cond):
    """Generate one narrative, skipping the call if a cached result exists."""
    path = cond["_path"]
    if path.exists():
        return {"status": "cached", "key": cond["_key"]}

    block = block_index.get(
        (cond["domain"], cond["model"], cond["xai"], cond["case_id"])
    )
    if block is None:
        return {
            "status": "missing_evidence",
            "key": cond["_key"],
            "error": f"no evidence block for {cond['domain']}/{cond['model']}/"
                     f"{cond['xai']}/case {cond['case_id']}",
        }

    system = PROMPT_TEXT[cond["prompt"]]
    user = build_user_message(block)

    text, tin, tout, fingerprint, error = call_with_retry(cond, system, user)
    if error is not None:
        return {"status": "failed", "key": cond["_key"], "error": error}

    record = {
        "block": cond["block"],
        "domain": cond["domain"],
        "model": cond["model"],
        "xai": cond["xai"],
        "llm": cond["llm"],
        "prompt": cond["prompt"],
        "prompt_hash": PROMPT_HASH[cond["prompt"]],
        "temperature": cond["temperature"],
        "seed": cond["seed"],
        "system_fingerprint": fingerprint,
        "case_id": cond["case_id"],
        "repeat": cond["repeat"],
        "explanation": text,
        "evidence": block["drivers"],
        "predicted_probability": block["predicted_probability"],
        "base_value": block["base_value"],
        "input_tokens": tin,
        "output_tokens": tout,
        "api_name": LLMS[cond["llm"]]["api_name"],
        "timestamp": pd.Timestamp.now().isoformat(),
    }
    path.write_text(json.dumps(record, ensure_ascii=False, indent=1), encoding="utf-8")
    return {"status": "generated", "key": cond["_key"]}


# [Cell 12] Batch execution
# Blocks are run one at a time so that partial progress is interpretable and a
# failure in one block does not obscure the state of the others.
def run_block(block_name, conds, max_workers=MAX_WORKERS):
    """Execute every pending condition in a block, in parallel."""
    pending = [c for c in conds if not c["_path"].exists()]
    print(f"[{block_name}] {len(conds):,} conditions, {len(pending):,} pending")

    if not pending:
        return Counter({"cached": len(conds)})

    outcomes = Counter()
    failures = []
    t0 = time.time()

    with ThreadPoolExecutor(max_workers=max_workers) as pool:
        futures = {pool.submit(run_condition, c): c for c in pending}
        for fut in tqdm(as_completed(futures), total=len(futures), desc=block_name):
            res = fut.result()
            outcomes[res["status"]] += 1
            if res["status"] in ("failed", "missing_evidence"):
                cond = futures[fut]
                failures.append({**{k: v for k, v in cond.items()
                                    if not k.startswith("_")},
                                 "error": res.get("error")})

    elapsed = time.time() - t0
    print(f"[{block_name}] {dict(outcomes)} in {elapsed / 60:.1f} min")

    if failures:
        fail_path = LOG_DIR / f"07_failures_{block_name}.json"
        fail_path.write_text(
            json.dumps(failures, ensure_ascii=False, indent=1), encoding="utf-8"
        )
        print(f"[{block_name}] {len(failures)} failures logged to {fail_path.name}")

    return outcomes


by_block = defaultdict(list)
for c in conditions:
    by_block[c["block"]].append(c)

BLOCK_ORDER = ["D", "D0", "A", "A_prime", "B", "C"]   # cheapest first
all_outcomes = {}

for block_name in BLOCK_ORDER:
    all_outcomes[block_name] = run_block(block_name, by_block[block_name])

print("\nOverall:", dict(sum(all_outcomes.values(), Counter())))


# [Cell 13] Retry pass for transient failures
# Rerun anything still missing after the main pass. Persistent failures usually
# indicate a configuration problem rather than a transient one.
still_missing = [c for c in conditions if not c["_path"].exists()]
print(f"Conditions still missing: {len(still_missing):,}")

if still_missing:
    print("Running retry pass with reduced concurrency")
    retry_outcomes = Counter()
    with ThreadPoolExecutor(max_workers=max(2, MAX_WORKERS // 3)) as pool:
        futures = {pool.submit(run_condition, c): c for c in still_missing}
        for fut in tqdm(as_completed(futures), total=len(futures), desc="retry"):
            retry_outcomes[fut.result()["status"]] += 1
    print("Retry outcomes:", dict(retry_outcomes))

    final_missing = [c for c in conditions if not c["_path"].exists()]
    if final_missing:
        print(f"\nWARNING: {len(final_missing):,} conditions could not be generated.")
        miss_df = pd.DataFrame([
            {k: v for k, v in c.items() if not k.startswith("_")}
            for c in final_missing
        ])
        save_table(
            miss_df.groupby(["block", "llm"]).size().reset_index(name="N missing"),
            "tab28_missing_generations",
        )
        print("Missing conditions are excluded from analysis and must be "
              "reported as incomplete cells.")


# [Cell 14] Consolidate the corpus
def load_corpus():
    """Read every cached generation into a single frame."""
    rows = []
    for path in tqdm(sorted(GEN_DIR.rglob("*.json")), desc="loading"):
        rows.append(json.loads(path.read_text(encoding="utf-8")))
    return pd.DataFrame(rows)


corpus = load_corpus()
print(f"Corpus size: {len(corpus):,} generations")

# Evidence is stored as a nested list; keep it as JSON text inside the parquet
corpus["evidence_json"] = corpus["evidence"].apply(
    lambda x: json.dumps(x, ensure_ascii=False)
)
corpus_out = corpus.drop(columns=["evidence"])
corpus_out.to_parquet(DATA_DIR / "generation_corpus.parquet", index=False)
print("Saved: generation_corpus.parquet")


# [Cell 15] Completeness audit
# Verifies that the corpus matches the plan cell by cell. Any shortfall is a
# threat to the balanced design and must be visible before analysis begins.
plan_counts = (
    plan.groupby(["block", "domain", "model", "xai", "llm", "prompt",
                  "temperature", "case_id"])
    .size().rename("planned").reset_index()
)
actual_counts = (
    corpus.groupby(["block", "domain", "model", "xai", "llm", "prompt",
                    "temperature", "case_id"])
    .size().rename("actual").reset_index()
)

audit = plan_counts.merge(
    actual_counts,
    on=["block", "domain", "model", "xai", "llm", "prompt", "temperature", "case_id"],
    how="left",
)
audit["actual"] = audit["actual"].fillna(0).astype(int)
audit["shortfall"] = audit["planned"] - audit["actual"]

incomplete = audit[audit["shortfall"] > 0]
print(f"Cells planned: {len(audit):,}")
print(f"Cells incomplete: {len(incomplete):,} "
      f"({len(incomplete) / len(audit) * 100:.2f}%)")

completeness = (
    audit.groupby("block")
    .agg(
        cells=("planned", "size"),
        planned=("planned", "sum"),
        actual=("actual", "sum"),
        incomplete_cells=("shortfall", lambda s: int((s > 0).sum())),
    )
    .reset_index()
)
completeness["completion_rate"] = (
    completeness["actual"] / completeness["planned"]
).round(4)

save_table(completeness, "tab29_corpus_completeness")
display(completeness)


# [Cell 16] Output quality checks
# Screens for degenerate generations that would distort claim extraction:
# truncated responses, refusals, or outputs that leaked the technical framing.
REFUSAL_MARKERS = [
    "i cannot", "i can't", "i'm unable", "i am unable",
    "as an ai", "i do not have enough",
]
LEAK_MARKERS = ["shap", "lime", "feature importance", "attribution value"]

corpus["n_chars"] = corpus["explanation"].str.len()
corpus["n_words"] = corpus["explanation"].str.split().str.len()
corpus["is_empty"] = corpus["explanation"].str.strip().eq("")
corpus["is_refusal"] = corpus["explanation"].str.lower().apply(
    lambda t: any(m in t for m in REFUSAL_MARKERS)
)
corpus["has_leak"] = corpus["explanation"].str.lower().apply(
    lambda t: any(m in t for m in LEAK_MARKERS)
)
corpus["likely_truncated"] = (
    corpus["output_tokens"] >= MAX_TOKENS - 5
) | ~corpus["explanation"].str.rstrip().str.endswith((".", "!", "?", ")"))

quality = (
    corpus.groupby(["block", "llm"])
    .agg(
        n=("explanation", "size"),
        mean_words=("n_words", "mean"),
        empty=("is_empty", "sum"),
        refusal=("is_refusal", "sum"),
        leaked=("has_leak", "sum"),
        truncated=("likely_truncated", "sum"),
    )
    .round(2)
    .reset_index()
)

save_table(quality, "tab30_generation_quality_checks")

flagged = corpus[
    corpus["is_empty"] | corpus["is_refusal"] | corpus["likely_truncated"]
]
print(f"Flagged generations: {len(flagged):,} "
      f"({len(flagged) / len(corpus) * 100:.2f}%)")

if len(flagged) / len(corpus) > 0.02:
    print("NOTE: flagged rate exceeds 2 per cent. Inspect examples below and "
          "decide whether to regenerate or to exclude with justification.")
    print(flagged[["block", "llm", "n_words", "explanation"]].head(5).to_string())

display(quality)


# [Cell 17] Exact-duplicate rate
# A direct precursor to the deep repeatability analysis. Identical text across
# repetitions is the strongest possible form of content stability; its absence
# under tau = 0 is the finding that Block D-0 exists to establish.
dup_rows = []
deep = corpus[corpus["block"].isin(["D", "D0"])]

for (block, llm, case_id), g in deep.groupby(["block", "llm", "case_id"]):
    texts = g["explanation"].str.strip().tolist()
    dup_rows.append({
        "block": block,
        "llm": llm,
        "case_id": case_id,
        "n_generations": len(texts),
        "n_unique": len(set(texts)),
        "unique_rate": len(set(texts)) / len(texts) if texts else np.nan,
    })

duplicates = pd.DataFrame(dup_rows)

dup_summary = (
    duplicates.groupby(["block", "llm"])
    .agg(
        n_cases=("case_id", "nunique"),
        mean_unique_rate=("unique_rate", "mean"),
        fully_identical_cases=("n_unique", lambda s: int((s == 1).sum())),
        fully_unique_cases=("unique_rate", lambda s: int((s == 1.0).sum())),
    )
    .round(4)
    .reset_index()
)

save_table(dup_summary, "tab31_exact_duplicate_rate")
display(dup_summary)


# [Cell 18] Cost and usage accounting
usage_rows = []
for llm, spec in LLMS.items():
    sub = corpus[corpus["llm"] == llm]
    if len(sub) == 0:
        continue
    usage_rows.append({
        "LLM": llm,
        "Vendor": spec["vendor"],
        "Generations": len(sub),
        "Input tokens": int(sub["input_tokens"].sum()),
        "Output tokens": int(sub["output_tokens"].sum()),
        "Mean output tokens": round(float(sub["output_tokens"].mean()), 1),
    })

usage = pd.DataFrame(usage_rows)
usage.loc[len(usage)] = {
    "LLM": "Total", "Vendor": "",
    "Generations": usage["Generations"].sum(),
    "Input tokens": usage["Input tokens"].sum(),
    "Output tokens": usage["Output tokens"].sum(),
    "Mean output tokens": None,
}

save_table(usage, "tab32_token_usage")
display(usage)


# [Cell 19] Figure: corpus composition
fig, axes = plt.subplots(1, 2, figsize=(7.6, 3.0))

block_counts = corpus["block"].value_counts().reindex(BLOCK_ORDER).dropna()
axes[0].bar(
    block_counts.index, block_counts.values,
    color="0.55", edgecolor="0.25", linewidth=0.7, width=0.6,
)
for i, v in enumerate(block_counts.values):
    axes[0].text(i, v, f"{int(v):,}", ha="center", va="bottom",
                 fontsize=7, color="0.2")
axes[0].set_ylabel("Generations")
axes[0].set_xlabel("Block")
axes[0].tick_params(axis="x", labelsize=8)

llm_counts = corpus["llm"].value_counts()
shades = ["0.25", "0.45", "0.62", "0.78"][:len(llm_counts)]
axes[1].bar(
    range(len(llm_counts)), llm_counts.values,
    color=shades, edgecolor="0.25", linewidth=0.7, width=0.6,
)
axes[1].set_xticks(range(len(llm_counts)))
axes[1].set_xticklabels(llm_counts.index, fontsize=7, rotation=20, ha="right")
axes[1].set_ylabel("Generations")

plt.tight_layout()
save_fig(fig, "fig20_corpus_composition")
plt.show()


# [Cell 20] Figure: explanation length by generator and prompt
fig, axes = plt.subplots(1, 2, figsize=(7.8, 3.2))

llms = list(corpus["llm"].unique())
box_data = [corpus[corpus["llm"] == m]["n_words"].dropna() for m in llms]
bp = axes[0].boxplot(box_data, widths=0.55, patch_artist=True, showfliers=False)
for patch, s in zip(bp["boxes"], ["0.25", "0.45", "0.62", "0.78"][:len(llms)]):
    patch.set_facecolor(s)
    patch.set_edgecolor("0.2")
    patch.set_linewidth(0.7)
for element in ["whiskers", "caps", "medians"]:
    for item in bp[element]:
        item.set_color("0.2")
        item.set_linewidth(0.8)
axes[0].set_xticklabels(llms, fontsize=7, rotation=20, ha="right")
axes[0].set_ylabel("Words per explanation")

prompts_present = [p for p in PROMPTS if p in set(corpus["prompt"])]
box_data = [corpus[corpus["prompt"] == p]["n_words"].dropna() for p in prompts_present]
bp = axes[1].boxplot(box_data, widths=0.55, patch_artist=True, showfliers=False)
for patch, s in zip(bp["boxes"], ["0.75", "0.52", "0.30"][:len(prompts_present)]):
    patch.set_facecolor(s)
    patch.set_edgecolor("0.2")
    patch.set_linewidth(0.7)
for element in ["whiskers", "caps", "medians"]:
    for item in bp[element]:
        item.set_color("0.2")
        item.set_linewidth(0.8)
axes[1].set_xticklabels(prompts_present, fontsize=8)
axes[1].set_ylabel("")

plt.tight_layout()
save_fig(fig, "fig21_explanation_length")
plt.show()


# [Cell 21] Figure: exact duplication under the two temperature settings
fig, ax = plt.subplots(figsize=(5.6, 3.0))

deep_llms = sorted(dup_summary["llm"].unique())
x = np.arange(len(deep_llms))
width = 0.35

for i, (block, label, shade) in enumerate(
    [("D", "tau = 1.0", "0.62"), ("D0", "tau = 0", "0.28")]
):
    vals = [
        dup_summary[
            (dup_summary["block"] == block) & (dup_summary["llm"] == m)
        ]["mean_unique_rate"].mean()
        for m in deep_llms
    ]
    ax.bar(
        x + (i - 0.5) * width, vals, width,
        color=shade, edgecolor="0.2", linewidth=0.7, label=label,
    )
    for xi, v in zip(x + (i - 0.5) * width, vals):
        if not np.isnan(v):
            ax.text(xi, v, f"{v:.2f}", ha="center", va="bottom",
                    fontsize=7, color="0.2")

ax.set_xticks(x)
ax.set_xticklabels(deep_llms, fontsize=7, rotation=20, ha="right")
ax.set_ylabel("Share of unique outputs")
ax.set_ylim(0, 1.1)
ax.legend(fontsize=7, frameon=True, loc="lower right")

plt.tight_layout()
save_fig(fig, "fig22_exact_duplication_by_temperature")
plt.show()


# [Cell 22] Persist the corpus manifest
# The manifest records what was actually generated, as distinct from what was
# planned. All downstream notebooks read the manifest rather than re-deriving
# the design from configuration.
manifest = {
    "notebook": "07_generation",
    "timestamp": pd.Timestamp.now().isoformat(),
    "design": FINAL,
    "llms": LLMS,
    "reference_conditions": {
        "llm": REFERENCE_LLM,
        "prompt": REFERENCE_PROMPT,
        "model": REFERENCE_MODEL,
        "xai": REFERENCE_XAI,
    },
    "prompt_hashes": PROMPT_HASH,
    "temperature": {"main": TEMPERATURE_MAIN},
    "seed": {"fixed": SEED_FIXED, "free": SEED_FREE},
    "max_tokens": MAX_TOKENS,
    "planned_total": len(plan),
    "generated_total": int(len(corpus)),
    "completeness": completeness.to_dict(orient="records"),
    "flagged_rate": round(float(len(flagged) / len(corpus)), 5),
    "token_usage": usage.to_dict(orient="records"),
}

with open(DATA_DIR / "corpus_manifest.json", "w", encoding="utf-8") as f:
    json.dump(manifest, f, ensure_ascii=False, indent=2, default=str)
print("Saved: corpus_manifest.json")

stamp = pd.Timestamp.now().strftime("%Y%m%d")
with open(LOG_DIR / f"07_run_{stamp}.json", "w", encoding="utf-8") as f:
    json.dump(manifest, f, ensure_ascii=False, indent=2, default=str)
print(f"Saved: logs/07_run_{stamp}.json")


# [Cell 23] Verification and handover notes
print("Outputs:")
for p in sorted(TAB_DIR.glob("tab2[7-9]*")) + sorted(TAB_DIR.glob("tab3[0-2]*")) \
        + sorted(FIG_DIR.glob("fig2[0-2]*")):
    print(f"  {p.relative_to(ROOT)}")

print(f"""
Corpus summary
--------------
Planned      {len(plan):,}
Generated    {len(corpus):,}
Completion   {len(corpus) / len(plan) * 100:.2f} per cent
Flagged      {len(flagged):,} ({len(flagged) / len(corpus) * 100:.2f} per cent)

Cached files under data/generations are the authoritative record. Re-running
this notebook regenerates nothing that already exists; delete the relevant
block directory to force regeneration.

Before proceeding, confirm that tab29_corpus_completeness shows no block
below 99 per cent. Incomplete cells unbalance the design and must be reported.

Next step: 08_content_distance.ipynb
""")
```
